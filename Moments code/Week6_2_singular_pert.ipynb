```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.linalg as la
from scipy.integrate import solve_ivp
```

```python
alpha = 1.0
beta = 0.5
A11 = np.array([[0., 1.],
                [-1., 0.]])
A12 = np.array([[0. ,0.],
                [0.,-beta]])
A21 = np.array([[0.,0.],
                [-alpha, alpha * beta]])
A22 = np.array([[0., 3.0],
                [-3., 0.]])

C1 = np.array([0., 1.0])
C2 = np.array([0., 1.0])

A22inv = la.inv(A22)

A_red = A11 - (A12 @ A22inv @ A21)
C_red = (C1 - C2 @ A22inv @ A21)
A_full = np.block([[A11, A12],
                    [A21, A22]])

eigvalA = la.eigvals(A_full)
eigvalA

#la.eigvals(A22)
```

```python
epsilon_list = np.linspace(0.001, 0.011, 10)
t_span = (0., 8.0*np.pi)
t_eval = np.linspace(t_span[0], t_span[1], 10000)
state0 = np.array([1.0, 1.0, 2.0, 0.5])

def full_system(t, state, epsilon):
    x = state[:2]
    z = state[2:]
    dx = A11 @ x + A12 @ z
    dz = (A21 @ x + A22 @ z) / epsilon

def reduced_state(t, x):
    dx = A_red @ x
    return dx

for epsilon in epsilon_list: 
    full_state_sol = solve_ivp(
        lambda t, state: full_system(t, state, epsilon),
        t_span, 
        state0, 
        t_eval = t_eval,
        method = 'Radau'
    )
```
